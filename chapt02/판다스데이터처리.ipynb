## 판다스 데이터 처리

### 데이터 정제

- 결측값(치)와 이상치를 제거하고 분석가능한 깨끗한 데이터로 수정하는 작업
    - 결측값 : NULL, None, Nan 등 입력되지 않은 필요 데이터. 나이 컬럼에 값 없음
    - 이상치 : 컬럼에 알맞지 않은 데이터. 점수 999점
    - 중복값 : 중복된 데이터. 같은 학생이 두 번 등록
    - 타입오류 : 일치하지 않은 데이터타입으로 입력된 값. 점수에 문자열로 저장. '구십구점'
    - 불필요한 공백 : 필요없는 스페이스키 입력. '   홍길동  '
    - 도메인오류 : 표현방식이 일치하지 않음. 'M', 'Male', '남자', '남성'... 
- ↑ 위 문제들을 해결하지 않으면 통계(계산) 불가능

### 데이터 전처리

- 데이터 정제를 하는 과정. 분석 프로젝트의 40~50% 차지
- AI 시대에서는 정제 과정이 상당 축소될 것 예상
- 사람의 검수 반드시 필요. 시간 대부분 여기에서 소요

```python
# 평균값 다시 ㄱ하기
age_mean = int(df['age'].mean().round())

df['age'].fillna(age_mean)  # -> 대입된게 아님. 보여지기만 한거
df['age'] = df['age'].fillna(age_mean)
```

- 결측치 채울 때 평균을 사용할 경우, 중앙값, 그 외 다른 값을 사용할 수도 있음

#### 2. 점수 이상치

```python
(df['score'] < 0) | (df['score'] > 100)
```

- score 컬럼이 object 타입으로 >, < 등의 비교연산 불가
- 숫자형 데이터로 변환

```python
# score 컬럼을 숫자형으로 변경, 만약 변경이 안되면 NaN으로 변경

df['score'] = pd.to_numeric(df['score'], errors='coerce')
```

```python
df
```

```python
# 점수 이상치 찾기
df[(df['score'] < 0) | (df['score'] > 100)]
```

```python
# 점수 이상치 -> NaN
df.loc[(df['score'] < 0) | (df['score'] > 100), 'score'] = np.nan
```

```python
df['score'].describe()
```

```python
df['score'] = [df['score'].isnull()]
```

```python
# 중앙값
score_median = df['score'].median()
```

```python
df['score'] = df['score'].fillna(score_median)
```

#### 이메일 형식 검사

- 정규표현식(Regular Expression)
    - r"^[\w\.-]+@[\w\.-]+\.\w+$" (이메일 포맷) 의 규칙에 맞는 데잍어가 이메일 타입이라는 의미

```python
pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"
```

```python
df[~(df['email'].str.match(pattern, na=False))]
```

```python
# 변수로 만들면 로직 처리 구문이 간결
valid_email = df['email'].str.match(pattern, na=False)
```

```python
df[~valid_email]
```
