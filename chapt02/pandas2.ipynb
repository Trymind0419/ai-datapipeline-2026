#### 행 선택, 인덱스행(loc)과 위치값행(iloc) 차이

````python
df2 = pd.DataFrame(raw_data,
                   index=['S0001', 'S0002', 'S0003', 'S0005', 'S0009', 'S0010'])

df2
````

````python
df2.loc['S0001' : 'S0009']
````

````python
# 파이썬 배열 인덱스와 동일
df2.iloc[ : 4]
````

- loc는 부여받은 인덱스로 행 선택 및 슬라이싱
- iloc는 원래부터의 위치값(파이썬 배열의 인덱스)과 동일

#### 조건 조회

- SQL WHERE절과 동일

````python
df['score'] >= 90
````

````python
df[df['score'] >= 90]
````

- NumPy와 동일한 조건 연산
- 파이썬 : and, or, not 사용
- NumPy, Pandas : `%`, `|`, `~` 사용
- 연산자 우선순위 문제로 각 조건은 반드시 `()` 사용할 것~

````python
df[(df['score'] > 90) & (df['age'] >= 21)]
````

````python
df[(df['score'] > 90) | (df['age'] <= 21)]
````

#### 컬럼 추가
- 사용하던 DF 이외에 필요한 컬럼이 생길 때 추가

````python
df['score_bonus'] = df['score'] + 5

df
````

#### 컬럼 수정
- 기존 컬럼 수정

````python
# 행선택 에서 score_bonus 점수가 100점이 넘는 행만 선택 후 100점으로 수정
df.loc[df['score_bonus']>100, 'score_bonus'] = 100

df
````

#### 컬럼 삭제
- 불필요한 컬럼은 삭제

````python
df.drop(columns=['score_bonus']) # 지워진 것처럼만 출력됨.

df
````

````python
df.drop(columns=['score_bonus'], inplace=True) # 두번 실행시 에러남

df
````

### 실습
- 상품데이터 활용

````python
data = {
    'product' : ['키보드', '마우스', '모니터', 'USB', '노트북'],
    'category' : ['주변기기', '주변기기', '디스플레이', '저장장치', '컴퓨터'],
    'price' : [30000, 15000, 250000, 12000, 1200000],
    'stock' : [10, 30, 5, 50, 3]
}

df_products = pd.DataFrame(data)
````

1. 전체 데이터 출력
2. product, price 컬럼 출력
3. 가격이 50000원 이상인 상품만 출력
4. 재고가 10개 이상인 상품만 출력
5. total_value 컬럼 추가, 계산식은 price * stock
6. 전체 재고의 합계 출력
7. 카테고리가 주변기기인 상품만 출력
