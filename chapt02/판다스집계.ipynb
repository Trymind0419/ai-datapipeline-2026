## 판다스 집계

### 집계

- 계산, 통계, 그룹핑으로 원데이터에서 수치적으로 필요한 값을 재가공ㅁ

```python
import pandas as pd
import numpy as np
```

```python
orders = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6],
    "customer": ["홍길동", "김철수", "이영희", "홍길동", "박민수", "김철수"],
    "product_id": ["P001", "P002", "P001", "P003", "P002", "P003"],
    "quantity": [2, 1, 3, 1, 2, 4],
    "unit_price": [30000, 15000, 30000, 8000, 15000, 8000],
    "order_date": [
        "2026-09-01",
        "2026-09-01",
        "2026-09-02",
        "2026-09-02",
        "2026-09-03",
        "2026-09-03"
    ]
})

### 날짜는 DB에서는 TimeStamp나 DateTime 이지만, 파이썬에서는 문자열로 씀

orders
# 주문번호, 고객명, 상품ID, 수량, 단가, 주문일자
```

#### 파생 컬럼 생성

- 단가 * 수량 -> 주문금액 컬럼

```python
orders['total_price'] = orders['quantity'] * orders['unit_price']
```

```python
orders
```

- 조건 연산자(>, <. =) 와 논리 연산자(&, |, ~) 등 사용해서 필요한 조건에 맞는 데이터만 필터링
- 필터링 후 파생컬럼 생성

```python
orders['customer_type'] = '일반'

orders
```

```python
orders['total_price'] >= 50000
```

```python
orders.loc[condition]
```

```python
orders.loc[condition, 'customer_type'] = 'VIP'
```

```python
orders
```

#### apply() 함수 사용

- apply() 함수 내에 사용할 사용자 정의 함수 생성
- 학점 등 여러 구분으로 나눠지는 도메인 값의 컬럼을 생성할 때 유용

```python
def get_order_grade(total_price):
    if total_price >= 80000:
        return 'A'
    elif total_price >= 30000:
        return 'B'
    else:
        return 'C'
```

```python
get_order_grade(25000)
```

```python
# apply 라는 판다스 함수. get_order_grade 라는 함수에서 total_price 파라미터를 자동으로 넣어줌.
# total_price 컬럼의 값을 사용해서 get_order_grade() 함수를 각 행별로 호출해서 grade 컬럼을 생성
orders['grade'] = orders['total_price'].apply(get_order_grade)
```

```python
orders
```

```python
orders.info()
```

```python
orders['order_date'] = pd. to_datetime(orders['order_date'])
```

```python
orders.info()
```

```python
# 날짜 형식으로 변경할 시 년, 월, 일, 시, 분, 초 분리하여 데이터 사용 가능
orders['order_date'].dt.year
orders['order_date'].dt.month
orders['order_date'].dt.day
```

#### CONCAT

- 데이터 합치기

```python
week1 = pd.DataFrame({
    "order_id": [1, 2],
    "customer": ["홍길동", "김철수"],
    "amount": [60000, 15000]
})

week2 = pd.DataFrame({
    "order_id": [3, 4],
    "customer": ["이영희", "박민수"],
    "amount": [90000, 30000]
})
```

```python
#이전 인덱스가 그대로 반영
pd.concat([week1, week2])
```

```python
# 일반적으로 가장 많이 사용하는 방법
total = pd.concat([week1, week2], ignore_index=True)

total
```

#### Merge
- DB Join과 동일

```python
products = pd.DataFrame({
    "product_id": ["P001", "P002", "P003"],
    "product_name": ["키보드", "마우스", "USB"],
    "category": ["주변기기", "주변기기", "저장장치"]
})
```

```python
orders
# 보통 product_id로 작성함. for 데이터 무결성 
```

```python
pd.merge(orders, products,on='product_id')
```

- how: inner, left, right 가 99% 쓰임

left: 왼쪽에 있는거는 다

right: 오른쪽에 있는거 다

```python
pd.merge(orders, products,on='product_id', how='left')
```

```python
products.loc[3] = {'product_id' : 'P004', 'product_name' : '모니터' , 'category' : '디스플레이'}
```

```python
products
```

```python
pd.merge(orders, products,on='product_id', how='right')
```

#### group by 집계

- DB group by 동일

```python
orders.groupby('customer')['total_price'].sum()
```

```python
# 데이터프레임으로 변환
orders.groupby('customer')['total_price'].sum().reset_index()
```

#### 여러 통계 한번에 계산

```python
# 고객별로 기본통계
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min', 'std']
)
```

```python
# 통계치 csv 파일 저장
orders.groupby('customer')['total_price'].agg(
    ['count', 'sum', 'mean', 'max', 'min', 'std']
).to_csv('../data/고객별구매통계.csv', encoding='utf8')
```
