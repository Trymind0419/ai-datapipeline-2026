#### Matplotlib 기본사용법

##### 선 그래프(Line Chart)

```python
plt.figure()
plt.plot([1, 2, 3, 4], [10,20,15,30])  # plot(`x축`, `y축`)
plt.show
```

```python
plt.figure(figsize=(20,5))  # figsize = (그래프너비, 그래프높이)
plt.plot([1, 2, 3, 4], [10,20,15,30])  # plot(`x축`, `y축`)
plt.title('선 그래프')
plt.xlabel('가로')
plt.ylabel('세로')
plt.show
```

##### Seaborn 패키지

- Matplotlib을 좀 더 예쁘게 출력 지원
- Seaborn용 추가 차트 기능도 존재

```python
sns.set_theme(style='darkgrid', font='Malgun Gothic')
```

```python
plt.figure(figsize=(20,5))  # figsize = (그래프너비, 그래프높이)
plt.plot([1, 2, 3, 4], [10,20,15,30])  # plot(`x축`, `y축`)
plt.title('선 그래프')
plt.xlabel('가로')
plt.ylabel('세로')
plt.show
```

```python
plt.figure(figsize=(10,5))  # figsize = (그래프너비, 그래프높이)
sns.lineplot(
    x=[1, 2, 3, 4],
    y=[10,20,15,30],
    marker='x',
    markersize=5,
    linewidth=2.5,
    color='#7f97b7' #RGB
    )  
plt.title('선 그래프', fontsize=16, fontweight='bold')
plt.xlabel('가로')
plt.ylabel('세로')

sns.despine() # 필요없음. 테두리 깔끔용
plt.show
```

##### 막대그래프

- 항목별(카테고리) 크기를 비교할 때 사용

```python
df.groupby('category')['total_price'].sum().reset_index()
```

```python
cate_sales = df.groupby('category')['total_price'].sum().reset_index()
```

```python
#x축
cate_sales['category']
```

```python
#y축
cate_sales['total_price']
```

```python
plt.figure(figsize=(10,5))

# 막대그래프 plt.bar() / (x, height)
plt.bar(x=cate_sales['category'], height=cate_sales['total_price'])
plt.title('항목별 판매금액', fontsize = 15, fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain', axis='y') # y축은 판매금액 그대로 출력
plt.show
```

##### 선 그래프
- 날짜별 매출 계산

```python
df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
daily_sales=df.groupby('order_date')['total_price'].sum().reset_index()
```

```python
plt.figure(figsize=(20,5))
plt.plot(daily_sales['order_date'], daily_sales['total_price'], 
         marker='o', 
         markersize=10,
         linewidth=5,
         color='#000000')

plt.title('일자별 판매금액', fontsize=20, fontweight='bold')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표시
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].to_string('%y%m%d')
)
plt.show()
```

#### 히스토그램
- 표준편차, 분포와 관계있는 그래프
- 숫자 데이터의 분포를 확인할 때 사용

```python
plt.figure(figsize=(10,5))

plt.hist(df['total_price'], bins=5) #구간 개수
plt.title('구매금액 분포도')
plt.xlabel('구매금액 분포')
plt.ylabel('갯수', rotation=0, labelpad=25)

plt.ticklabel_format(style='plain', axis='x')

plt.show()
```

##### 산점도
- 두 숫자 데이터간의 관계를 확인 그래프
- 데이터 분석 시 가장 많이 사용

```python
plt.figure(figsize=(10,5))

#산점도
plt.scatter(x=df['quantity'], y=df['total_price'])
plt.title('구매수량, 금액간의 관계도')
plt.xlabel('구매갯수')
plt.ylabel('구매분포', rotation=0, labelpad=25)

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

##### 박스플롯
- 데이터의 이상

```python
plt.figure(figsize=(10,5))

#산점도
plt.boxplot(df['total_price'])
plt.title('구매액 분포와 이상치', fontsize=15)
plt.ylabel('구매금액', rotation=0, labelpad=25)

plt.ticklabel_format(style='plain', axis='y')
plt.show()
```

##### 선 그래프 마커위치에 데이터 표시

```python
plt.figure(figsize=(10,5))
plt.plot(daily_sales['order_date'], daily_sales['total_price'], 
         marker='o', 
         markersize=10,
         linewidth=5,
         color="#0000005C")

# 마커위치에 실제 데이터 표시
for x, y in zip(daily_sales['order_date'], daily_sales['total_price']):
    plt.text(
        x,y,
        f"{y:,}", # y값 = 금액
        ha='left', # 가로축 정렬 left, center, right - 실제 정렬과 이름 반대
        va='bottom' # 세로축 정렬 top, center, bottom - 실제 정렬과 이름 반대
    )

plt.title('일자별 판매금액', fontsize=20, fontweight='bold')
plt.ylabel('판매금액')
plt.ticklabel_format(style='plain', axis='y')

# 실제 데이터가 있는 위치의 x좌표 레이블 표시
# plt.xticks(
#     daily_sales['order_date'],
#     daily_sales['order_date'].to_string('%y%m%d')
# )
plt.xticks(
    daily_sales['order_date'],
    daily_sales['order_date'].dt.strftime('%y-%m-%d')
)
plt.show()
```

#### Seaborn으로 손쉽게 차트 구현

- Maplotlib은 판다스 DF를 바로 사용 못함.
- Seaborn은 DF를 매개변수로 활용가능
- Matplotlib은 x,y(height) 값을 시리즈로 만들어서 전달
- Seaborn은 데이터프레임 그대로 data 속성에 전달

```python
plt.figure(figsize=(10,5))

# 막대그래프 plt.bar() -> Seaborn barplot으로 변경
sns.barplot(data=cate_sales, x='category', y='total_price')
plt.title('항목별 판매금액', fontsize = 15, fontweight='bold')
plt.xlabel('카테고리')
plt.ylabel('판매금액')

plt.ticklabel_format(style='plain', axis='y') # y축은 판매금액 그대로 출력
plt.show
```

##### Seaborn Heatmap
- 숫자 데이터 분포를 색상으로 표현하는 그래프

```python
# 구매수량, 단가, 구매금액 간의 상관관계
df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
corr = df[['quantity', 'unit_price', 'total_price']].corr()
```

```python
plt.figure(figsize=(10,5))

# 히트맵
sns.heatmap(data=corr, annot=True, cmap='Blues')
plt.title('구매수량과 단가 간의 상관관계')
plt.show
```
