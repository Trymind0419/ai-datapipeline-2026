````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(int(price*0.9)) #10% 할인

discount_prices
````

- 결측치 정리 방법 : 결측치는 제거하거나(건수 감소), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기(최소값, 최대값, 중앙값 ...)


#### 이상치 핸들링

- 일반적인 범위를 벗어나는 비정상 데이터.
- 예로 부산 1월 일별 평균기온: -90도

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0])
temperatures
````

````python
valid_temps = temperatures[
    (~np.isnan(temperatures)) &     #nan은 무조건 제거
    (temperatures >= 15.0 ) &       # 최저기온이 15 아래로 내려갈 수 없으므로 제거
    (temperatures <= 50)            # 최고기온이 50도 이상 올라갈 수 없어 제거
]
valid_temps
````

- 전처리(PreProcessing) - 분석 이전에 이상치, 결측치를 없애는 작업

### 실습

#### 기온데이터 전처리

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0, 24.0, 23.5, np.nan])
temperatures
````

- 결측값 개수 출력
- 결측값, 이상치 제거
- 정상 데이터 출력
- 평균 기온 출력
- 최고 기온 출력
- 최저 기온 출력

````python
# 결측값 출력
np.isnan(temperatures)
````

````python
# 결측값 갯수
print(np.isnan(temperatures).sum())
missing_count = np.isnan(temperatures).sum()
````

````python
# 결측값, 이상치 제거 및 정상 데이터 출력
valid_temps = temperatures[
    (~np.isnan(temperatures)) &
    (temperatures >= -15.0) &
    (temperatures <= 50.0)
]

valid_temps
````

````python
valid_temps.mean()
valid_temps.max()
valid_temps.min()
````

### 매출 데이터 전처리

- 결측값 개수 출력
- 결측값 제외
- 0원 10만원 내의 정상 매출만 남김
- 정상 매출의 평균, 최대값, 최소값 출력
- 정상 매출에서 평균이상인 값만 출력

````python
# 파이썬 리스트에서 None 대신 np.nan으로 결측치를 입력하면 NumPy 패키지 연산 가능
raw_sales = [12000, 15000, np.nan, 30000, -5000, 25000, 9999999, 18000, np.nan, 22000]

raw_sales
````

````python
# NumPy 배열로 변경
np_raw_sales = np.array(raw_sales)

np_raw_sales
````

````python
# 결측값 확인
np.isnan(raw_sales)
````

````python
np.isnan(raw_sales).sum()
````

````python
np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]
````

````python
valid_sales = np_raw_sales[
    (~np.isnan(np_raw_sales)) &
    (np_raw_sales >= 0) &
    (np_raw_sales <= 100000)
]

valid_sales
````

````python
mean_sales = valid_sales.mean()
valid_sales.min()
valid_sales.max()
````

````python
# 평균 매출액보다 높은 매출 찾기
valid_sales[valid_sales >= mean_sales]
````
