### Python 리스트(배열) 과 NumPy 배열 비교

- 파이썬에 기존 리스트(배열) 자료구조 존재. 속도적면에서 NumPy가 우세

#### 파이썬으로 배열작업

````python
prices = [10000, 15000, 12000, 30000, 25000] # 5개 가격이 든 배열
discount_prices = []

for price in prices:
    discount_prices.append(int(price*0.9)) #10% 할인

discount_prices
````

#### NumPy 배열 작업

````python

np_prices = np.array(prices) # 넘파이 배열로 변경
np_discount_prices = np_prices * 0.9   # for 문 쓸 필요 X, 리스트 항목 별 0.9 곱해줌
np_discount_prices
````

- Numpy 배열은 반복문을 사용하지 않아도 배열 전체에 계산을 수행할 수 있음. `벡터화 연산`

#### NumPy 데이터타입
- int8 : -128 ~ 127 (총 256)
- int16 : -32768 ~ 32767 (총 65536)
- int32 : 32비트 정수
- int64 : 64비트 정수(Numpy 기본 정수)
- uint8 : 0 ~ 255 / unsigned : - 부호가 없는
- uint16 : 0 ~ 65535
- float16 : 실수
- float32: 32비트 실수
- float64: 64비트 실수(Numpy 기본 실수)
- complex64 : 복소수
- complex128 : 128비트 복소수
- bool : True/False
- str_ : 문자열
- object_ : 객체

#### 인덱싱 / 슬라이싱

- 파이썬 배열에서 기능과 유사

````python
# 인덱싱
print(data[1])
print(data[-1])

# 슬라이싱
print(data[:]) # 배열전체
print(data[1:4]) # 1부터 3까지
print(data[:2])
print(data[3:])
````

#### 조건 필터링
- SQL WHERE 절과 동일, 조건에 맞는 데이터만 골라내기

````python
print(prices)
np_prices
````

````python
prices >= 20000  ## 넘파이 배열이 아니라 파이썬 배열이라 안됨
````

````python
np_prices >= 20000
````

````python
condition = np_prices >= 20000
condition
````

````python
# np_prices에서 True인 값들만 출력
np_prices[condition]
np_prices[np_prices >= 20000]
````

````python
datas = np.array([10, -1, 80, 999, 30, 120])
datas
````

````python
# & : and, | : or
datas[(datas >= 0) & (datas <= 100)]
````
